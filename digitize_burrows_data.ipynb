```python
import numpy as np
```

```python
# From crop, the (x,y) coordinates of data points are:
x_y_vals = [
    (106,1036),
    (160,1042),
    (266,1042),
    (322,1014),
    (374,988),
    (374,1017),
    (427,1015),
    (481,988),
    (540,962),
    (540,913),
]
```

```python
# Simple linear scale, crop is 540px x 1240px,
# and range is [-20,0] x [0,50]. So conversion is:
def theta_r_vals(x,y):
    R_meters_hr = 50*(1240-y)/1240
    R_meters_min = R_meters_hr/60
    theta = (-20/540) * (540 - x)
    return (theta, R_meters_min)

theta_r_data = [
    theta_r_vals(x,y) for (x,y) in x_y_vals
]
```

```python
# There are 2 values with |degrees| < 1. Sullivan et al. appear to have averaged
# them and used that for normalization.
R0 = np.mean([R for (theta,R) in theta_r_data if np.absolute(theta) < 1])
data = [(theta, R/R0) for (theta, R) in theta_r_data]
print(data)
```

```python
with open("py_out/burrows_1994_data.csv", "w") as f:
    f.write("theta_deg,R_normalized\n")
    for (theta,R) in data:
        f.write(f"{theta:.2f},{R:.4f}\n")
```
