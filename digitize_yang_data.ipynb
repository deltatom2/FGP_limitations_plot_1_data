```python
# Explicit values for the relative rate of spread are given in Yang et al. (2018), 
# so we can use them without any processing. Duplicate instances of (0, 1.000) have been omitted.
data = [
    (0, 1.000),
    (-10, 1.006),
    (-10, 0.955),
    (-20, 0.905),
    (-20, 0.895),
    (-30, 0.980),
    (-30, 0.942),

    (-10, 1.061),
    (-10, 1.004),
    (-20, 0.947),
    (-20, 0.998),
    (-30, 1.181),
    (-30, 1.154),
]

with open("py_out/yang_2018_data.csv", "w") as f:
    f.write("theta_deg,R_normalized\n")
    for (theta,R) in data:
        f.write(f"{theta:.2f},{R:.4f}\n")
```
