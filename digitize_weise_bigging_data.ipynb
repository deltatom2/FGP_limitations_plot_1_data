```python
import numpy as np
```

```python
# Weise and Biging explicitly provide their values, as (slope_percent, R_meter_sec), in 5 categories:
# "high heading" (HH), "low heading" (LH), "no wind", "low backing" (LB) and "low heading" (LH).
# 
# Sullivan et al. appear to have used only a subset of the values. I've chosen to include
# them all for convenience.

def normalize(R, R0):
    return R/R0

def percent_to_deg(percent_slope):
    return (180/np.pi) * np.arctan(percent_slope)

def convert_all(vals, R0):
    return [(percent_to_deg(p), normalize(R,R0)) for (p,R) in vals]

HB_0_slope_R_val = 1.45
HB_slope_percent_vals = [
    (-0.3, 1.58),
    (-0.15, 1.00),
    (0, 1.45),
]
HB_data = convert_all(HB_slope_percent_vals, HB_0_slope_R_val)

LB_0_slope_R_val = 2.30
LB_slope_percent_vals = [
    (-0.3, 1.40),
    (-0.15, 1.65),
    (0, 2.30),
]
LB_data = convert_all(LB_slope_percent_vals, LB_0_slope_R_val)

NW_0_slope_R_val = 2.68
NW_slope_percent_vals = [
    (-0.3, 1.55),
    (-0.15, 3.30),
    (0, 2.68),
]
NW_data = convert_all(NW_slope_percent_vals, NW_0_slope_R_val)

LH_0_slope_R_val = 7.75
LH_slope_percent_vals = [
    (-0.3, 5.00),
    (-0.15, 4.30),
    (0, 7.75),
]
LH_data = convert_all(LH_slope_percent_vals, LH_0_slope_R_val)

HH_0_slope_R_val = 30.70
HH_slope_percent_vals = [
    (-0.3, 22.43),
    (-0.15, 20.25),
    (0, 30.70),
]
HH_data = convert_all(HH_slope_percent_vals, HH_0_slope_R_val)

data = HB_data + LB_data + NW_data + LH_data + HH_data
```

```python
with open("py_out/weise_biging_1997_data.csv", "w") as f:
    f.write("theta_deg,R_normalized\n")
    for (theta,R) in data:
        f.write(f"{theta:.2f},{R:.4f}\n")
```
