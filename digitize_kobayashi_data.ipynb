```python
# Crop of graph used was 593px x 850px
# Visible axis ticks on figure 1 occur at:
# (850 px, 0.1), (591px, 0.5), (480px, 1.0), (110px, 10.0)
# Clearly a logarithmic scale, so we try and fit y = A + B log(R)
import numpy as np

y_ticks = np.array([850, 591, 480, 110])
R_ticks = np.array([0.1, 0.5, 1.0, 10.0])
R_ticks_log = np.log(R_ticks)

X = np.column_stack([np.ones_like(R_ticks_log), R_ticks_log])
result = np.linalg.lstsq(X, y_ticks, rcond=None)
A, B = result[0]              # argmin over b of ||X b - y||_2
print(A,B)

```

Hence relation seems to be
$$R \approx e^{(480-y)/160}.$$
$\theta$-axis requies no guesswork: all points plotted clearly lie on particular $\theta$-ticks.

```python
def R(y):
    a = (480-y)/160
    return np.exp(a)
```

```python
# By eye, y values of data points appear to be:
# (680px, -35 deg), (660px, -25 deg), (610px, 0 deg)
y_theta_coords = [
    (-35, 680),
    (-25, 660),
    (0, 610)
]
R_theta_coords = [(theta, R(y)) for (theta,y) in y_theta_coords]
print(R_theta_coords)
```

```python
# The R value at 0 degrees is 0.4437
R0 = 0.4437
data = [(theta,R/R0) for (theta,R) in R_theta_coords]

with open("py_out/kobayashi_1991_data.csv", "w") as f:
    f.write("theta_deg,R_normalized\n")
    for (theta,R) in data:
        f.write(f"{theta:.2f},{R:.4f}\n")
```
