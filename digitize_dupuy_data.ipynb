```python
import numpy as np
```

```python
# 3 y values per theta value in figure. From crop of figure, the values are [written as (theta;y,y,y)], where y is pixel value of from crop and theta is read from the axis ticks:
#  (0;191,243,255), (-10;195,204,236), (-20;316,323,330), (-30;334,347,353)
theta_y_vals = [
    [0, [191,243,255]],
    [-10, [195,204,236]],
    [-20, [316,323,330]],
    [-30, [334,347,353]]
]
```

```python
# Height of cropped graph is 530px, so the conversion is:
def R(y):
    R_mm_sec = 12 * (530-y)/530
    R_m_min = (60/100) * R_mm_sec
    return R_m_min

theta_r_vals = [
    (theta, [R(y) for y in y_vals]) 
    for (theta,y_vals) in theta_y_vals
]
```

```python
R0 = np.mean(theta_r_vals[0][1])
data = [(theta,[R/R0 for R in R_vals]) for (theta,R_vals) in theta_r_vals]
```

```python
with open("py_out/dupuy_1995_data.csv", "w") as f:
    f.write("theta_deg,R_normalized\n")
    for (theta,R_vals) in data:
        for R in R_vals:
            f.write(f"{theta:.2f},{R:.4f}\n")
```
